```python
import plotly.figure_factory as ff
import numpy as np
import math
```

```python
def g(x, y):
    if x > y:
        return math.sqrt(x - y)
    return 0.0
```

```python
def gg(x, y):
    return x + y, x - y
```

```python
X, Y = np.meshgrid([0.0, 0.5, 1.0], [0.0, 0.5, 1.0])
```

```python
gv = np.vectorize(g)
Z = gv(X, Y)
```

```python
ggv = np.vectorize(gg)
```

```python
ggv(X, Y)
```
